### CatNinja API

```python
import requests

api_url = "https://catfact.ninja/fact"
response = requests.get(api_url)
data = response.json()
data

df = pd.DataFrame([data])
df.to_csv("Exports/CatNinja.csv",index=False)
```

```python
import requests
import pandas as pd
import time

api_url = "https://catfact.ninja/fact"

facts = []

for i in range(10):
    response = requests.get(api_url)
    data = response.json()
    facts.append(data)
    print(f"Fetched fact {i+1}: {data}")
    time.sleep(1)

df = pd.DataFrame(facts)

df
```
