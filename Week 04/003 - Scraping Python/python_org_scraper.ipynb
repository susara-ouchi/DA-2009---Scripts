### Scraping Python.org

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.python.org/"

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
```

```python
news_section = soup.select_one("div.blog-widget")
```

```python
news_items = news_section.select("ul.menu li")
```

```python
news_data = []

for item in news_items:
    
    date = item.select_one("time").get_text(strip=True)
    
    title = item.select_one("a").get_text(strip=True)
    
    url = item.select_one("a").get("href")
    
    news_data.append([date, title, url])
```

```python
news_df = pd.DataFrame(
    news_data,
    columns=["Date", "Title", "URL"]
)

news_df
```

```python
news_df.to_csv("Latest_Python_News.csv", index=False)
```
