```python
from bs4 import BeautifulSoup
import pandas as pd
```

Importing file

```python
file_path = "Sample_portfolio/index.html"

with open(file_path,'r',encoding='utf-8') as file:
    html = file.read()
```

Create the soup

```python
soup = BeautifulSoup(html,'html.parser')
print(soup.prettify())
```

# Extract nav items

```python
nav_items = soup.select("nav .nav-links li a")
nav_items
```

```python
for i in nav_items:
    dict_items = {"Section":i.text.strip(), "Link":i.get("href")}
    print(dict_items)
```

```python
nav_items = soup.select("nav .nav-links li a")
nav_data = [{"Section":link.text.strip(), "Link":link.get("href")} for link in nav_items]
nav_data
```

```python
nav_df = pd.DataFrame(nav_data)
nav_df
```

### The fun begins here!

Profile

```python
name = soup.select_one(".left-head h1").get_text(strip=True)

subtitle = soup.select_one(".subtitle").get_text(strip=True)

print(name)
print(subtitle)
```

Contact

```python
contact = soup.select_one(".contact")

contact_text = contact.get_text("\n", strip=True)

print(contact_text)
```

```python
contact_items = []

for p in contact.find_all("p"):
    contact_items.append(p.get_text(" ", strip=True))

contact_items
```

About

```python
about_section = soup.find("div", id="about")

about_text = about_section.find("p").get_text(" ", strip=True)

print(about_text)
```

Education

```python
education_section = soup.find("div", id="education")

education = []

for item in education_section.find_all("li"):
    education.append(item.get_text(" ", strip=True))

education
```

Projects

```python
projects_section = soup.find("div", id="projects")

projects = []

for item in projects_section.find_all("li"):
    projects.append(item.get_text(" ", strip=True))

projects
```

```python
def split_list(input_list, splitter=":"):
    first_half = []
    last_half = []

    for input_item in input_list:
        first_half_1, last_half_1 = input_item.split(splitter, 1)

        first_half.append(first_half_1.strip())
        last_half.append(last_half_1.strip())

    return first_half, last_half
split_list(projects)
```

Skills

```python
skills_section = soup.find("div", id="skills")

skills = []

for item in skills_section.find_all("div", class_="info-box"):
    skills.append(item.get_text(" ", strip=True))

skills
```

Set up a df

```python
cv_data = []

# Profile
cv_data.append({
    "section": "Profile",
    "subsection": "Name",
    "content": name
})

cv_data.append({
    "section": "Profile",
    "subsection": "Title",
    "content": subtitle
})

# About
cv_data.append({
    "section": "Profile",
    "subsection": "About",
    "content": about_text
})

# Contact
for item in contact_items:
    cv_data.append({
        "section": "Contact",
        "subsection": "",
        "content": item
    })

# Education
for item in education:
    cv_data.append({
        "section": "Education",
        "subsection": "",
        "content": item
    })

# Projects
for item in projects:
    cv_data.append({
        "section": "Projects",
        "subsection": "",
        "content": item
    })

# Skills
for item in skills:
    cv_data.append({
        "section": "Skills",
        "subsection": "",
        "content": item
    })
```

```python
cv_df = pd.DataFrame(cv_data)

cv_df
```

### Fun little side quest!

Let's make it neater

```python
# Function to split the text further
def split_list(items, separator):
    result = []

    for item in items:
        first, second = item.split(separator, 1)
        result.append([first.strip(), second.strip()])

    return result
```

```python
split_list(projects, ":")
```

```python
split_list(education, "–")
```

```python
split_list(contact_items,":")
```

```python
for item in split_list(contact_items,":"):
    print(item[0])
```

```python
cv_data = []

# Profile
cv_data.append({
    "section": "Profile",
    "subsection": "Name",
    "content": name
})

cv_data.append({
    "section": "Profile",
    "subsection": "Title",
    "content": subtitle
})

# About
cv_data.append({
    "section": "Profile",
    "subsection": "About",
    "content": about_text
})

# Contact
for item in split_list(contact_items,":"):
    cv_data.append({
        "section": "Contact",
        "subsection": item[0],
        "content": item[1]
    })

# Education
for item in split_list(education,"–"):
    cv_data.append({
        "section": "Education",
        "subsection": item[0],
        "content": item[1]
    })

# Projects
for item in split_list(projects, ":"):
    cv_data.append({
        "section": "Projects",
        "subsection": item[0],
        "content": item[1]
    })

# Skills
for item in split_list(skills, ":"):
    cv_data.append({
        "section": "Skills",
        "subsection": item[0],
        "content": item[1]
    })
```

```python
cv_df = pd.DataFrame(cv_data)

cv_df
```
